```python
from pathlib import Path

import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

```python
DATA_DIR = Path("../data")

TRAIN_PATH = DATA_DIR / "train_test.csv"
VALIDATION_PATH = DATA_DIR / "validation.csv"

train = pd.read_csv(TRAIN_PATH)
validation = pd.read_csv(VALIDATION_PATH)

print("Training shape:", train.shape)
print("Validation shape:", validation.shape)
```

```python
print("TRAINING")
print("Missing:", train["weight"].isnull().sum())
print("Negative:", (train["weight"] < 0).sum())

print()

print("VALIDATION")
print("Missing:", validation["weight"].isnull().sum())
print("Negative:", (validation["weight"] < 0).sum())
```

```python
train["weight_missing"] = train["weight"].isnull().astype(int)

validation["weight_missing"] = validation["weight"].isnull().astype(int)
```

```python
print(
    train["weight_missing"].value_counts()
)
```

```python
train["weight"] = train["weight"].abs()

validation["weight"] = validation["weight"].abs()
```

```python
print(
    "Training negative weights:",
    (train["weight"] < 0).sum()
)

print(
    "Validation negative weights:",
    (validation["weight"] < 0).sum()
)
```

```python
weight_median = train["weight"].median()

print("Training median weight:", weight_median)
```

```python
train["weight"] = train["weight"].fillna(weight_median)

validation["weight"] = validation["weight"].fillna(weight_median)
```

```python
validation["weight"].fillna(weight_median)
```

```python
validation["weight"].fillna(
    validation["weight"].median()
)
```

```python
print("Training missing:", train["weight"].isnull().sum())
print("Validation missing:", validation["weight"].isnull().sum())

print("Training negative:", (train["weight"] < 0).sum())
print("Validation negative:", (validation["weight"] < 0).sum())
```

```python
train["market_index_missing"] = (
    train["market_index"].isnull().astype(int)
)

validation["market_index_missing"] = (
    validation["market_index"].isnull().astype(int)
)
```

```python
print(
    train["market_index_missing"].value_counts()
)
```

```python
market_median = train["market_index"].median()

print("Training market median:", market_median)
```

```python
train["market_index"] = (
    train["market_index"].fillna(market_median)
)

validation["market_index"] = (
    validation["market_index"].fillna(market_median)
)
```

```python
print(
    "Training missing:",
    train["market_index"].isnull().sum()
)

print(
    "Validation missing:",
    validation["market_index"].isnull().sum()
)
```

```python
print(
    "Training missing quote signal:",
    train["quote_signal"].isnull().sum()
)

print(
    "Validation missing quote signal:",
    validation["quote_signal"].isnull().sum()
)
```

```python
train["date"] = pd.to_datetime(train["date"])

validation["date"] = pd.to_datetime(validation["date"])
```

```python
print(train["date"].dtype)
print(validation["date"].dtype)
```

```python
train["day_of_week"] = train["date"].dt.dayofweek

validation["day_of_week"] = validation["date"].dt.dayofweek
```

```python
train[
    ["date", "day_of_week"]
].head()
```

```python
train["month"] = train["date"].dt.month

validation["month"] = validation["date"].dt.month
```

```python
train["day_of_month"] = train["date"].dt.day

validation["day_of_month"] = validation["date"].dt.day
```

```python
train["week_of_year"] = (
    train["date"].dt.isocalendar().week.astype(int)
)

validation["week_of_year"] = (
    validation["date"].dt.isocalendar().week.astype(int)
)
```

```python
train["is_weekend"] = (
    train["day_of_week"] >= 5
).astype(int)

validation["is_weekend"] = (
    validation["day_of_week"] >= 5
).astype(int)
```

```python
train[
    [
        "date",
        "month",
        "week_of_year",
        "day_of_month",
        "day_of_week",
        "is_weekend"
    ]
].head(10)
```

```python
train["log_distance"] = np.log1p(train["distance"])

validation["log_distance"] = np.log1p(
    validation["distance"]
)
```

```python
train[
    [
        "distance",
        "log_distance"
    ]
].head(10)
```

```python
train["log_weight"] = np.log1p(train["weight"])

validation["log_weight"] = np.log1p(
    validation["weight"]
)
```

```python
train[
    [
        "weight",
        "log_weight",
        "weight_missing"
    ]
].head()
```

```python
train["lane"] = (
    train["pickup"]
    + " -> "
    + train["delivery"]
)

validation["lane"] = (
    validation["pickup"]
    + " -> "
    + validation["delivery"]
)
```

```python
train[
    [
        "pickup",
        "delivery",
        "lane"
    ]
].head()
```

```python
train["equipment"] = train["equipment"].astype("category")

validation["equipment"] = validation["equipment"].astype("category")
```

```python
print(train["equipment"].dtype)

print(
    train["equipment"].cat.categories
)
```

```python
pickup_categories = sorted(
    set(train["pickup"]) | set(validation["pickup"])
)

delivery_categories = sorted(
    set(train["delivery"]) | set(validation["delivery"])
)


train["pickup"] = pd.Categorical(
    train["pickup"],
    categories=pickup_categories
)

validation["pickup"] = pd.Categorical(
    validation["pickup"],
    categories=pickup_categories
)



train["delivery"] = pd.Categorical(
    train["delivery"],
    categories=delivery_categories
)

validation["delivery"] = pd.Categorical(
    validation["delivery"],
    categories=delivery_categories
)
```

```python
train_lanes = set(train["lane"])


train["new_lane"] = 0

validation["new_lane"] = (
    ~validation["lane"].isin(train_lanes)
).astype(int)


validation["lane"] = validation["lane"].where(
    validation["lane"].isin(train_lanes),
    "UNKNOWN_LANE"
)


lane_categories = sorted(train_lanes) + ["UNKNOWN_LANE"]

train["lane"] = pd.Categorical(
    train["lane"],
    categories=lane_categories
)

validation["lane"] = pd.Categorical(
    validation["lane"],
    categories=lane_categories
)
```

```python
train["log_posted_rate"] = np.log(
    train["posted_rate"]
)
```

```python
train[
    [
        "posted_rate",
        "log_posted_rate"
    ]
].head()
```

```python
'''posted_rate
    │
    │ log()
    ↓
log_posted_rate
    │
    │ MODEL
    ↓
prediction
    │
    │ exp()
    ↓
predicted_rate'''
```

```python
FEATURES = [
    # Freight
    "distance",
    "log_distance",
    "weight",
    "log_weight",
    "weight_missing",
    "equipment",

    # Geography
    "pickup",
    "delivery",
    "pickup_lat",
    "pickup_lon",
    "delivery_lat",
    "delivery_lon",
    "lane",

    # Market
    "market_index",
    "market_index_missing",
    "quote_signal",

    # Time
    "month",
    "week_of_year",
    "day_of_month",
    "day_of_week",
    "is_weekend"
]

TARGET = "log_posted_rate"
```

```python
X = train[FEATURES]

y = train[TARGET]

X_validation = validation[FEATURES]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Validation X shape:", X_validation.shape)
```

```python
print("Training missing values:")

print(
    X.isnull().sum()
)
```

```python
print("Validation missing values:")

print(
    X_validation.isnull().sum()
)
```

```python
print("FINAL FEATURE ENGINEERING CHECK")
print("=" * 50)

print("Training rows:", len(X))
print("Validation rows:", len(X_validation))

print()

print("Number of features:", len(FEATURES))

print()

print("Target missing:", y.isnull().sum())

print()

print(
    "Training infinite values:",
    np.isinf(
        X.select_dtypes(include=np.number)
    ).sum().sum()
)

print(
    "Validation infinite values:",
    np.isinf(
        X_validation.select_dtypes(include=np.number)
    ).sum().sum()
)
```

```python
print("Training missing features:")
print(X.isnull().sum()[X.isnull().sum() > 0])

print("\nValidation missing features:")
print(X_validation.isnull().sum()[X_validation.isnull().sum() > 0])
```

```python
PROCESSED_DIR = DATA_DIR / "processed"

PROCESSED_DIR.mkdir(exist_ok=True)


train.to_csv(
    PROCESSED_DIR / "train_processed.csv",
    index=False
)

validation.to_csv(
    PROCESSED_DIR / "validation_processed.csv",
    index=False
)

print("Processed data saved successfully.")
```

```python
print(
    "Train saved:",
    (PROCESSED_DIR / "train_processed.csv").exists()
)

print(
    "Validation saved:",
    (PROCESSED_DIR / "validation_processed.csv").exists()
)
```
